## Constants

### Data

```python
# Dataset in-repo: ./dataset/.../train|valid and ./dataset/test/test/*.jpg
_PROJECT_ROOT = os.getcwd()
BASE_DIR = os.path.join(
    _PROJECT_ROOT,
    "dataset",
    "New Plant Diseases Dataset(Augmented)",
    "New Plant Diseases Dataset(Augmented)",
)
TRAIN_DIR = os.path.join(BASE_DIR, "train")
VAL_DIR = os.path.join(BASE_DIR, "valid")
TEST_DIR = os.path.join(_PROJECT_ROOT, "dataset", "test", "test")
if not os.path.isdir(TRAIN_DIR):
    raise FileNotFoundError(
        f"Train folder not found at {TRAIN_DIR}. Set the notebook working directory to the project root."
    )
if not os.path.isdir(TEST_DIR):
    raise FileNotFoundError(f"Test folder not found at {TEST_DIR}")
IMG_SHAPE = (256, 256)
BATCH_SIZE = 64
```

### Model

```python
INPUT_DEPTH_MULTIPLIER = 8
BASE_FILTER_COUNT = 32
DENSE_NODES = 512
DROPOUT_PERCENTS = [0.2, 0.2]
```

### Training

```python
EPOCHS = 30
PATIENCE = 5
QUEUE_SIZE = 25
TB_LOGS = "tensorboard_logs/fit/" + datetime.datetime.now().strftime("%Y%m%d-%H%M%S")
PLOTS_DPI = 200
```

## Data Loading

### Data Distribution

#### Train

```python
total_train = 0

for d in os.listdir(TRAIN_DIR):
    path = os.path.join(TRAIN_DIR, d)
    img_count = len(os.listdir(path))
    total_train += img_count
    print(f'{d:50} : {img_count}')

print(f'\nTotal training images: {total_train}')
```

#### Validation

```python
total_val = 0

for d in os.listdir(VAL_DIR):
    path = os.path.join(VAL_DIR, d)
    img_count = len(os.listdir(path))
    total_val += img_count
    print(f'{d:50} : {img_count}')

print(f'\nTotal validation images: {total_val}')
```

#### Test

```python
test_image_data = []
test_image_filenames = []

for img_name in os.listdir(TEST_DIR):
    img = load_img(os.path.join(TEST_DIR, img_name), target_size = IMG_SHAPE)
    test_image_data.append(img_to_array(img, dtype = 'uint8'))
    test_image_filenames.append(img_name)
    print(img_name)
    
test_image_data = np.array(test_image_data)/255
print(f'\nTotal testing images: {len(test_image_data)}')
```

### Image Generators

#### Train

```python
train_image_gen = ImageDataGenerator(
        rescale = 1./255,
        rotation_range = 15,
        width_shift_range = 0.05,
        height_shift_range = 0.05,
        shear_range = 0.05,
        zoom_range = 0.05,
        horizontal_flip = True,
        vertical_flip = True,
        fill_mode = 'reflect'
    )

train_data_gen = train_image_gen.flow_from_directory(
        batch_size = BATCH_SIZE,
        directory = TRAIN_DIR,
        shuffle = True,
        target_size = IMG_SHAPE
    )
```

```python
plot_images(train_data_gen)
```

#### Validation

```python
val_image_gen = ImageDataGenerator(rescale = 1./255)

val_data_gen = val_image_gen.flow_from_directory(
        batch_size = BATCH_SIZE,
        directory = VAL_DIR,
        shuffle = True,
        target_size = IMG_SHAPE
    )
```

```python
plot_images(val_data_gen)
```

### Class lookup dictionary

```python
total_classes = len(os.listdir(TRAIN_DIR))

print(f'\nTotal number of classes: {total_classes}')
```

```python
class_name_lookup = {name: index for index, name in train_data_gen.class_indices.items()}

for k, v in class_name_lookup.items():
    print(f"{k:2} : {v}")
```

## Model

### Input Layer

```python
inputs = tf.keras.Input(shape = (*IMG_SHAPE, 3), name = "Input")
inputs.shape
```

### Depthwise Convolution + Pointwise Convolution + Max Pooling blocks

#### Block 1

```python
depth_conv_1 = DepthwiseConv2D(2, depth_multiplier = INPUT_DEPTH_MULTIPLIER, padding = "same", activation = "relu", name = "DwC_1")(inputs)
depth_conv_1.shape
```

```python
point_conv_1 = Conv2D(BASE_FILTER_COUNT, 1, activation = "relu", name = "PwC_1")(depth_conv_1)
point_conv_1.shape
```

```python
max_pool_1 = MaxPooling2D(name = "MP_1")(point_conv_1)
max_pool_1.shape
```

#### Block 2

```python
depth_conv_2 = DepthwiseConv2D(2, padding = "same", activation = "relu", name = "DwC_2")(max_pool_1)
depth_conv_2.shape
```

```python
point_conv_2 = Conv2D(BASE_FILTER_COUNT * 2, 1, activation = "relu", name = "PwC_2")(depth_conv_2)
point_conv_2.shape
```

```python
max_pool_2 = MaxPooling2D(name = "MP_2")(point_conv_2)
max_pool_2.shape
```

#### Block 3

```python
depth_conv_3 = DepthwiseConv2D(2, padding = "same", activation = "relu", name = "DwC_3")(max_pool_2)
depth_conv_3.shape
```

```python
point_conv_3 = Conv2D(BASE_FILTER_COUNT * 4, 1, activation = "relu", name = "PwC_3")(depth_conv_3)
point_conv_3.shape
```

```python
max_pool_3 = MaxPooling2D(name = "MP_3")(point_conv_3)
max_pool_3.shape
```

#### Block 4

```python
depth_conv_4 = DepthwiseConv2D(2, padding = "same", activation = "relu", name = "DwC_4")(max_pool_3)
depth_conv_4.shape
```

```python
point_conv_4 = Conv2D(BASE_FILTER_COUNT * 8, 1, activation = "relu", name = "PwC_4")(depth_conv_4)
point_conv_4.shape
```

```python
max_pool_4 = MaxPooling2D(name = "MP_4")(point_conv_4)
max_pool_4.shape
```

#### Block 5

```python
depth_conv_5 = DepthwiseConv2D(2, padding = "same", activation = "relu", name = "DwC_5")(max_pool_4)
depth_conv_5.shape
```

```python
point_conv_5 = Conv2D(BASE_FILTER_COUNT * 16, 1, activation = "relu", name = "PwC_5")(depth_conv_5)
point_conv_5.shape
```

```python
max_pool_5 = MaxPooling2D(name = "MP_5")(point_conv_5)
max_pool_5.shape
```

### Global Average Pooling layers for each block

```python
global_avg_1 = GlobalAveragePooling2D(name = "GAP_1")(max_pool_1)
global_avg_1.shape
```

```python
global_avg_2 = GlobalAveragePooling2D(name = "GAP_2")(max_pool_2)
global_avg_2.shape
```

```python
global_avg_3 = GlobalAveragePooling2D(name = "GAP_3")(max_pool_3)
global_avg_3.shape
```

```python
global_avg_4 = GlobalAveragePooling2D(name = "GAP_4")(max_pool_4)
global_avg_4.shape
```

```python
global_avg_5 = GlobalAveragePooling2D(name = "GAP_5")(max_pool_5)
global_avg_5.shape
```

### Concatenation Layer

```python
concatenated = Concatenate(name = "Concat")([global_avg_1, global_avg_2, global_avg_3, global_avg_4, global_avg_5])
concatenated.shape
```

### Dropout and Dense Layers

```python
dropout_1 = Dropout(DROPOUT_PERCENTS[0], name = "Droput_1")(concatenated)
dropout_1.shape
```

```python
dense_1 = Dense(DENSE_NODES, activation = "relu", name = "Dense_1")(dropout_1)
dense_1.shape
```

```python
dropout_2 = Dropout(DROPOUT_PERCENTS[1], name = "Dropout_2")(dense_1)
dropout_2.shape
```

```python
dense_2 = Dense(total_classes, activation = "softmax", name = "Output")(dropout_2)
dense_2.shape
```

### Model Summary

```python
model = Model(inputs = inputs, outputs = dense_2, name = "Plant_Leaf")

model.compile(optimizer = 'adam', 
              loss = 'categorical_crossentropy',
              metrics = ['accuracy'])
 
model.summary()
```

```python
try:
    plot_model(model, to_file="plots/model.jpg", show_shapes=True, dpi=PLOTS_DPI)
except Exception as e:
    print("plot_model skipped (install graphviz + pydot for diagrams):", e)
```

## Training

### Callbacks

```python
earlyStop = EarlyStopping(monitor = 'val_loss', patience = PATIENCE, restore_best_weights = True)
tensorBoard = TensorBoard(log_dir = TB_LOGS, histogram_freq = 1)
```

### History

```python
%%time

history = model.fit(
    train_data_gen,
    epochs = EPOCHS,
    validation_data = val_data_gen,
    max_queue_size = QUEUE_SIZE,
    callbacks = [earlyStop, tensorBoard]
)
```

```python
acc = history.history['accuracy']
val_acc = history.history['val_accuracy']

loss = history.history['loss']
val_loss = history.history['val_loss']

epochs_range = history.epoch

plt.figure(figsize = (16, 6))

plt.subplot(1, 2, 1)
plt.plot(epochs_range, acc, label = 'Training Accuracy')
plt.plot(epochs_range, val_acc, label = 'Validation Accuracy')
plt.legend(loc = 'lower right')
plt.title('Training and Validation Accuracy')

plt.subplot(1, 2, 2)
plt.plot(epochs_range, loss, label = 'Training Loss')
plt.plot(epochs_range, val_loss, label = 'Validation Loss')
plt.legend(loc = 'upper right')
plt.title('Training and Validation Loss')

plt.suptitle('Plant Leaf Disease Detection')

plt.savefig('plots/acc_and_loss.jpg', dpi = PLOTS_DPI, bbox_inches='tight')
plt.show()
```

## MobileNetV2 Model (Transfer Learning)

### Build Model

```python
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.layers import BatchNormalization

mobilenet_base = MobileNetV2(
    input_shape = (*IMG_SHAPE, 3),
    include_top = False,
    weights = "imagenet"
)

mobilenet_base.trainable = False

mn_inputs  = tf.keras.Input(shape = (*IMG_SHAPE, 3), name = "MN_Input")
mn_x       = mobilenet_base(mn_inputs, training = False)
mn_x       = GlobalAveragePooling2D(name = "MN_GAP")(mn_x)
mn_x       = BatchNormalization(name = "MN_BN")(mn_x)
mn_x       = Dropout(0.2, name = "MN_Dropout")(mn_x)
mn_outputs = Dense(total_classes, activation = "softmax", name = "MN_Output")(mn_x)

mobilenet_model = Model(inputs = mn_inputs, outputs = mn_outputs, name = "MobileNetV2_PlantLeaf")

mobilenet_model.compile(
    optimizer = "adam",
    loss = "categorical_crossentropy",
    metrics = ["accuracy"]
)

mobilenet_model.summary()
```

### Training

#### Callbacks

```python
mn_tb_logs     = "tensorboard_logs/fit/mobilenet_" + datetime.datetime.now().strftime("%Y%m%d-%H%M%S")
mn_earlyStop   = EarlyStopping(monitor = "val_loss", patience = PATIENCE, restore_best_weights = True)
mn_tensorBoard = TensorBoard(log_dir = mn_tb_logs, histogram_freq = 1)
```

#### History

```python
%%time

mn_history = mobilenet_model.fit(
    train_data_gen,
    epochs = EPOCHS,
    validation_data = val_data_gen,
    max_queue_size = QUEUE_SIZE,
    callbacks = [mn_earlyStop, mn_tensorBoard]
)
```

```python
mn_acc          = mn_history.history["accuracy"]
mn_val_acc      = mn_history.history["val_accuracy"]
mn_loss         = mn_history.history["loss"]
mn_val_loss     = mn_history.history["val_loss"]
mn_epochs_range = mn_history.epoch

plt.figure(figsize = (16, 6))

plt.subplot(1, 2, 1)
plt.plot(mn_epochs_range, mn_acc,     label = "Training Accuracy")
plt.plot(mn_epochs_range, mn_val_acc, label = "Validation Accuracy")
plt.legend(loc = "lower right")
plt.title("MobileNetV2 - Training and Validation Accuracy")

plt.subplot(1, 2, 2)
plt.plot(mn_epochs_range, mn_loss,     label = "Training Loss")
plt.plot(mn_epochs_range, mn_val_loss, label = "Validation Loss")
plt.legend(loc = "upper right")
plt.title("MobileNetV2 - Training and Validation Loss")

plt.suptitle("MobileNetV2 Plant Leaf Disease Detection")
plt.savefig("plots/mobilenet_acc_and_loss.jpg", dpi = PLOTS_DPI, bbox_inches = "tight")
plt.show()
```

```python
fig, axes = plt.subplots(2, 2, figsize = (18, 12))

axes[0, 0].plot(history.epoch, history.history["accuracy"],     label = "Train", color = "blue")
axes[0, 0].plot(history.epoch, history.history["val_accuracy"], label = "Val",   color = "blue", linestyle = "--")
axes[0, 0].set_title("Custom CNN - Accuracy")
axes[0, 0].legend()

axes[0, 1].plot(mn_history.epoch, mn_history.history["accuracy"],     label = "Train", color = "orange")
axes[0, 1].plot(mn_history.epoch, mn_history.history["val_accuracy"], label = "Val",   color = "orange", linestyle = "--")
axes[0, 1].set_title("MobileNetV2 - Accuracy")
axes[0, 1].legend()

axes[1, 0].plot(history.epoch, history.history["loss"],     label = "Train", color = "blue")
axes[1, 0].plot(history.epoch, history.history["val_loss"], label = "Val",   color = "blue", linestyle = "--")
axes[1, 0].set_title("Custom CNN - Loss")
axes[1, 0].legend()

axes[1, 1].plot(mn_history.epoch, mn_history.history["loss"],     label = "Train", color = "orange")
axes[1, 1].plot(mn_history.epoch, mn_history.history["val_loss"], label = "Val",   color = "orange", linestyle = "--")
axes[1, 1].set_title("MobileNetV2 - Loss")
axes[1, 1].legend()

plt.suptitle("Model Comparison: Custom CNN vs MobileNetV2", fontsize = 14)
plt.tight_layout()
plt.savefig("plots/comparison_curves.jpg", dpi = PLOTS_DPI, bbox_inches = "tight")
plt.show()
```

### Validation Accuracy & Loss Comparison

```python
custom_best_val_acc     = max(history.history["val_accuracy"])
mobilenet_best_val_acc  = max(mn_history.history["val_accuracy"])
custom_best_val_loss    = min(history.history["val_loss"])
mobilenet_best_val_loss = min(mn_history.history["val_loss"])

fig, axes = plt.subplots(1, 2, figsize = (12, 5))

models_labels = ["Custom CNN", "MobileNetV2"]
val_accs      = [custom_best_val_acc,  mobilenet_best_val_acc]
val_losses    = [custom_best_val_loss, mobilenet_best_val_loss]
colors        = ["steelblue", "darkorange"]

bars0 = axes[0].bar(models_labels, val_accs, color = colors)
axes[0].set_ylim([0.85, 1.0])
axes[0].set_title("Best Validation Accuracy")
axes[0].set_ylabel("Accuracy")
for bar, v in zip(bars0, val_accs):
    axes[0].text(bar.get_x() + bar.get_width()/2, v + 0.001,
                 f"{v:.4f}", ha = "center", fontweight = "bold")

bars1 = axes[1].bar(models_labels, val_losses, color = colors)
axes[1].set_title("Best Validation Loss")
axes[1].set_ylabel("Loss")
for i, v in enumerate(val_losses):
    axes[1].text(i, v + 0.001, f"{v:.4f}", ha = "center", fontweight = "bold")

plt.suptitle("Best Validation Metrics Comparison", fontsize = 14)
plt.tight_layout()
plt.savefig("plots/comparison_metrics.jpg", dpi = PLOTS_DPI, bbox_inches = "tight")
plt.show()
```
